```python
import requests
import pandas as pd

def fetch_tweet_texts(user_id, api_key, output_csv="tweets.csv"):
    tweet_texts = []
    url = "https://api.twitterapi.io/twitter/user/last_tweets"
    headers = {"X-API-Key": api_key}

    for _ in range(10):  
        params = {"userId": user_id}
        response = requests.get(url, headers=headers, params=params)

        if response.status_code != 200:
            print(f"reached 200 error --")
            break

        tweets = response.json().get("data", {}).get("tweets", [])
        if not tweets:
            break

        tweet_texts.extend([tweet["text"] for tweet in tweets])

    df = pd.DataFrame(tweet_texts, columns=["text"])
    df.to_csv(output_csv, index=False)
```
