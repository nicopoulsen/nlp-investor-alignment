```python
import requests
import pandas as pd

def fetch_tweet_texts(user_id, api_key, output_csv="tweets.csv", total_tweets=200):
    tweet_texts = []
    url = "https://api.twitterapi.io/twitter/user/last_tweets"
    headers = {"X-API-Key": api_key}
    cursor = ""  

    while len(tweet_texts) < total_tweets:
        params = {
            "userId": user_id,
            "cursor": cursor
        }

        response = requests.get(url, headers=headers, params=params)

        if response.status_code != 200:
            print(f"Error fetching tweets: {response.status_code}")
            break

        data = response.json()
        tweets = data.get("data", {}).get("tweets", [])
        tweet_texts.extend([tweet["text"] for tweet in tweets])

        # here we added next cursor argument as the API paginates 20 tweets per call
        cursor = data.get("next_cursor")
        
        if not cursor:
            print("Reached the end, no more cursor provided.")
            break

        print(f"Total tweets fetched: {len(tweet_texts)}")

    tweet_texts = tweet_texts[:total_tweets]

    df = pd.DataFrame(tweet_texts, columns=["text"])
    df.to_csv(output_csv, index=False)
    print(f"Saved {len(tweet_texts)} tweets to {output_csv}")


```
